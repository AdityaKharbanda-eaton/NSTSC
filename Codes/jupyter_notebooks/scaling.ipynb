```python
num_rows = 5  # number of different sinusoidal waveforms
num_samples = 400  # samples per waveform

# generate x-values from 0 to 2*pi
x_values = np.linspace(0, 6 * np.pi, num_samples)

# assign different phases for each row
phases = np.linspace(0, 2 * np.pi, num_rows, endpoint=False)

# create an array where each row is a sinusoidal waveform with a unique phase shift
waveforms = np.array([np.sin(x_values + phase) for phase in phases])
print(waveforms)
```

```python
import matplotlib.pyplot as plt

for waveform in waveforms:
    plt.plot(x_values, waveform)
plt.title("All Waveforms")
plt.xlabel("x values")
plt.ylabel("Amplitude")
plt.show()
```
