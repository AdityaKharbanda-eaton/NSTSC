```python
import numpy as np
import pandas as pd
import os
```

```python
class0_data = np.full((10,20), 50)
class1_data = np.full((10,20), 30)
print("class0_data shape:", class0_data.shape)
print("class1_data shape:", class1_data.shape)
```

```python
x_data = np.concatenate((class0_data, class1_data), axis=0)
labels = np.concatenate((np.zeros(10), np.ones(10)), axis=0)
labels = labels.reshape((20, 1))
print("x_data shape:", x_data.shape)
print("labels shape:", labels.shape)
```

```python
train_data = np.concatenate((labels, x_data), axis = 1)
print(train_data.shape)
```

```python
train_data
```

```python
x_test_data = np.concatenate((class1_data, class0_data), axis=0)
test_labels = np.concatenate((np.ones(10), np.zeros(10)), axis=0).reshape((20, 1))
test_data = np.concatenate((test_labels, x_test_data), axis = 1)
test_data.shape
```

```python
test_data
```

```python
train_df = pd.DataFrame(train_data)
test_df = pd.DataFrame(test_data)
print("Train DataFrame shape:", train_df.shape)
print("Test DataFrame shape:", test_df.shape)
```

```python
train_df = train_df.astype(int)
test_df = test_df.astype(int)
print("Train DataFrame dtypes:\n", train_df.dtypes)
print("Test DataFrame dtypes:\n", test_df.dtypes)
```
